Q1. NumPy Square Matrix
Create the following NumPy matrix:
[4 7
 2 6]

Write a program to:
1. Find the determinant.
2. Find the inverse.
3. Find the transpose.
4. Calculate the sum of all elements.
5. Display its shape and dimensions.

```python
import numpy as np

#crate the Numpy matrix
A = np.array([[4, 7],
              [2, 6]])

print("Matrix:")
print(A)

# 1. Find determinant
det = np.linalg.det(A)
print("\nDeterminant:", det)

# 2. Find inverse
inverse = np.linalg.inv(A)
print("\nInverse:")
print(inverse)

# 3. Find transpose
transpose = A.T
print("\nTranspose:")
print(transpose)

# 4. Sum of all elements
total_sum = np.sum(A)
print("\nSum of all elements:", total_sum)

# 5. Display shape and dimensions
print("\nShape:", A.shape)
print("Dimensions:", A.ndim)
```

Q2. NumPy Reshaping
Create a NumPy array containing numbers from 1 to 12.

Perform:
1. Display the original array.
2. Reshape it into a 3x4 matrix.
3. Reshape it into a 2x6 matrix.
4. Display the first row.
5. Display the last column.
6. Calculate the sum of each row.

```python
import numpy as np

# Create NumPy array containing numbers from 1 to 12
arr = np.arange(1, 13)

# 1. Display the original array
print("Original Array:")
print(arr)

# 2. Reshape into 3x4 matrix
matrix_3x4 = arr.reshape(3, 4)
print("\n3x4 Matrix:")
print(matrix_3x4)

# 3. Reshape into 2x6 matrix
matrix_2x6 = arr.reshape(2, 6)
print("\n2x6 Matrix:")
print(matrix_2x6)

# 4. Display the first row
print("\nFirst Row:")
print(matrix_3x4[0])

# 5. Display the last column
print("\nLast Column:")
print(matrix_3x4[:, -1])

# 6. Calculate the sum of each row
print("\nSum of Each Row:")
print(np.sum(matrix_3x4, axis=1))
```

Q3. Temperature Dictionary
Create the following dictionary:
temperature = {
  "Monday": 31,
  "Tuesday": 33,
  "Wednesday": 30,
  "Thursday": 35,
  "Friday": 32,
  "Saturday": 36,
  "Sunday": 34
}

Find:
1. Maximum temperature
2. Minimum temperature
3. Average temperature
4. Day with the highest temperature
5. Days having temperature above 33

```python
# Create the temperature dictionary
temperature = {
    "Monday": 31,
    "Tuesday": 33,
    "Wednesday": 30,
    "Thursday": 35,
    "Friday": 32,
    "Saturday": 36,
    "Sunday": 34
}

# 1. Maximum temperature
max_temp = max(temperature.values())
print("Maximum temperature:", max_temp)

# 2. Minimum temperature
min_temp = min(temperature.values())
print("Minimum temperature:", min_temp)

# 3. Average temperature
average_temp = sum(temperature.values()) / len(temperature)
print("Average temperature:", average_temp)

# 4. Day with the highest temperature
highest_day = max(temperature, key=temperature.get)
print("Day with the highest temperature:", highest_day)

# 5. Days having temperature above 33
above_33 = [day for day, temp in temperature.items() if temp > 33]
print("Days having temperature above 33:", above_33)
```

Q4. Dictionary Comprehension
Create a dictionary containing marks of six students.

Using dictionary comprehension, create a new dictionary containing only
students who scored 80 or above.

Also:
1. Display the original dictionary.
2. Display the filtered dictionary.
3. Count the number of selected students.

```python
# Create a dictionary containing marks of six students
marks = {
    "Rahul": 75,
    "Aman": 85,
    "Priya": 92,
    "Riya": 68,
    "Arjun": 80,
    "Neha": 88
}

# 1. Display the original dictionary
print("Original Dictionary:")
print(marks)

# Using dictionary comprehension to select students
# who scored 80 or above
selected_students = {
    name: mark
    for name, mark in marks.items()
    if mark >= 80
}

# 2. Display the filtered dictionary
print("\nFiltered Dictionary:")
print(selected_students)

# 3. Count the number of selected students
count = len(selected_students)
print("\nNumber of selected students:", count)
```

Q5. Subject-wise Marks Analysis Using Dictionary
Create the following dictionary containing marks of five students:
marks = {
  "Amit": {"Python": 78, "DBMS": 82, "Maths": 75},
  "Riya": {"Python": 91, "DBMS": 88, "Maths": 94},
  "Rahul": {"Python": 65, "DBMS": 70, "Maths": 68},
  "Sneha": {"Python": 84, "DBMS": 79, "Maths": 88},
  "Arjun": {"Python": 73, "DBMS": 76, "Maths": 80}
}

Write a Python program to:
1. Display the complete dictionary.
2. Calculate the average marks in Python.
3. Calculate the average marks in DBMS.
4. Calculate the average marks in Maths.
5. Find the subject having the highest average marks.

```python
# Create the dictionary containing marks
marks = {
    "Amit": {"Python": 78, "DBMS": 82, "Maths": 75},
    "Riya": {"Python": 91, "DBMS": 88, "Maths": 94},
    "Rahul": {"Python": 65, "DBMS": 70, "Maths": 68},
    "Sneha": {"Python": 84, "DBMS": 79, "Maths": 88},
    "Arjun": {"Python": 73, "DBMS": 76, "Maths": 80}
}

# 1. Display the complete dictionary
print("Complete Dictionary:")
print(marks)

# Calculate average marks for each subject
python_avg = sum(student["Python"] for student in marks.values()) / len(marks)
dbms_avg = sum(student["DBMS"] for student in marks.values()) / len(marks)
maths_avg = sum(student["Maths"] for student in marks.values()) / len(marks)

# 2. Average marks in Python
print("\nAverage marks in Python:", python_avg)

# 3. Average marks in DBMS
print("Average marks in DBMS:", dbms_avg)

# 4. Average marks in Maths
print("Average marks in Maths:", maths_avg)

# Store subject averages in a dictionary
subject_averages = {
    "Python": python_avg,
    "DBMS": dbms_avg,
    "Maths": maths_avg
}

# 5. Find the subject having the highest average marks
highest_subject = max(subject_averages, key=subject_averages.get)

print("Subject with highest average marks:", highest_subject)
```
